```python
import glob 
import pandas as pd

# 1. Grab the paths to all 8 CSV files in the npx folder
npx_files = glob.glob('raw_data/npx/*.csv')

# 2. Create an empty list to hold the data
npx_data_list = []

# 3. Loop through each file, read it, and add it to the list
for file in npx_files:
    df = pd.read_csv(file)
    npx_data_list.append(df)

# 4. Combine all datasets into one massive master dataframe
if len(npx_files) > 0:
    npx_master_df = pd.concat(npx_data_list, ignore_index=True)
    print(f"Successfully combined {len(npx_files)} files!")
else:
    print("Uh oh! No files were found. Let's double-check the folder path.")
```

```python
# --- PHASE 2.1: REMOVING HEAVILY MISSING COLUMNS ---

# We want to keep columns that have AT LEAST 80% real data (max 20% missing)
# Calculate what 80% of our total rows equals
minimum_valid_data = int(0.8 * len(npx_master_df))

# Drop the junk columns and save it as a new "clean" dataframe
npx_clean_df = npx_master_df.dropna(axis=1, thresh=minimum_valid_data)

print(f"Original number of protein columns: {npx_master_df.shape[1]}")
print(f"New number of protein columns: {npx_clean_df.shape[1]}")
print(f" We successfully dropped {npx_master_df.shape[1] - npx_clean_df.shape[1]} useless columns!")
```

```python
# --- PHASE 2.2: THE PROPER PROTEOMICS CLEANING METHOD ---

# Step 1: Only drop proteins that are "dead" (missing more than 90% of their data)
# We keep columns that have at least 10% real data
min_valid = int(0.1 * len(npx_master_df))
npx_saved_df = npx_master_df.dropna(axis=1, thresh=min_valid)

print(f"1. Dropped 'dead' proteins. Columns safely remaining: {npx_saved_df.shape[1]}")

# Step 2: For the surviving proteins, treat NaN as "Below Limit of Detection"
# We fill the blanks with the minimum recorded value for each specific protein
npx_imputed_df = npx_saved_df.fillna(npx_saved_df.min(numeric_only=True))

print("2. Missing values successfully filled with minimum baselines!")
print(f"Total missing values left in NPX data: {npx_imputed_df.isna().sum().sum()}")
```
